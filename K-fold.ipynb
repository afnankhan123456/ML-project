# K-fold - Cross Validation Technique 

```python
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
```

```python
dir(data)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data = data.data , columns= data.feature_names )
```

```python
df 
```

```python
df['Target'] = data.target
```

```python
df.info()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import KFold , cross_validate, cross_val_score
```

```python
x = df.drop(columns = 'Target')
y = df['Target']

x_tarin ,x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
print(len(x_tarin) , len(x_test) , len(y_train), len(y_test))
```

```python
s = StandardScaler()
```

```python
saclled_data = s.fit_transform(x_tarin)
```

```python
saclled_data
```

```python
model = RandomForestClassifier()
```

```python
cv = 5
```

```python
k = KFold(cv)
```

```python
score = cross_validate(model , saclled_data,y_train,cv = k)
```

```python
score
```

```python
score['test_score'].mean()
```

```python
from sklearn.model_selection import cross_val_score
```

```python
final_score = cross_val_score(model , saclled_data,y_train,cv = k)
```

```python
final_score.mean()
```
